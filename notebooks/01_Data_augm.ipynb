```python
import pandas as pd
import os
from pathlib import Path

# pep_folder = "NLFR_analysis"
# wt_seq = "NLFRRVWEL"
# pep_file = "TCRneo_duplex_fingerprint_data_WT_TCR.tsv"

# pep_folder = "EVDP_analysis"
# wt_seq = "EVDPIGHLY"
# pep_file = "TCRA3_duplex_fingerprint_data_WT_TCR.tsv"

pep_folder = "SLLQ_analysis"
wt_seq = "SLLQHLIGL"
pep_file = "PRAMEb_duplex_fingerprint_data_WT_TCR.tsv"
```

```python
tsv_path = f"../data/{pep_folder}/{pep_file}"
df = pd.read_csv(tsv_path, sep="\t")
```

```python
input_af3 = pd.DataFrame({
    'name': df['Peptide_name'],
    'peptide': df['Peptide_seq'],
    'TRA_aa': df['TRA_aa_trimmed'],
    'TRB_aa': df['TRB_aa_trimmed'],
    'allele': df['HLA']
})
```

```python
output_path = f"../data/{pep_folder}/af3_input_base3.csv"
input_af3.to_csv(output_path, index=False)

output_path_res = f"../results/{pep_folder}/af3_input_base3.csv"
input_af3.to_csv(output_path_res, index=False)
```

## Creating a peptide.fasta for the Inverse Folding analysis

```python
fasta_lines = []
for _, row in df.iterrows():
  header = row["Peptide_mut"]  
  seq = row["Peptide_seq"]  
  fasta_lines.append(f">{header}\n{seq}")

fasta_out = Path(f"../data/{pep_folder}/mutated_peptides.fasta")
fasta_out.write_text("\n".join(fasta_lines))
```
